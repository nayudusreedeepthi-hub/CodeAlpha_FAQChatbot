```python
# Simple E-Commerce Mini Project

products = {
    1: ("Laptop", 50000),
    2: ("Mobile", 20000),
    3: ("Headphones", 2000),
    4: ("Mouse", 800)
}

cart = []

while True:
    print("\n===== E-Commerce Store =====")
    print("1. View Products")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Checkout")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        print("\nProducts:")
        for pid, item in products.items():
            print(pid, "-", item[0], "- ₹", item[1])

    elif choice == "2":
        pid = int(input("Enter Product ID: "))
        qty = int(input("Enter Quantity: "))

        if pid in products:
            cart.append((products[pid][0], products[pid][1], qty))
            print("Product Added Successfully!")
        else:
            print("Invalid Product ID")

    elif choice == "3":
        total = 0
        print("\nYour Cart:")
        if len(cart) == 0:
            print("Cart is Empty")
        else:
            for item in cart:
                cost = item[1] * item[2]
                total += cost
                print(item[0], "x", item[2], "= ₹", cost)
            print("Total = ₹", total)

    elif choice == "4":
        total = sum(item[1] * item[2] for item in cart)
        print("\nFinal Bill = ₹", total)
        print("Thank you for shopping!")
        cart.clear()

    elif choice == "5":
        print("Program Ended")
        break

    else:
        print("Invalid Choice")
```
